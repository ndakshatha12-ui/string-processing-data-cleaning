## A small text-cleaning function

```python
def clean_text(s):
    return " ".join(s.strip().split()).lower()

print(clean_text("   Messy   Text   Here  "))
```

## Cleaning names

```python
def clean_name(n):
    return " ".join(n.strip().split()).title()

print(clean_name("  raj   kumar  singh "))
```

```python
records = [" Asha Kumar ", "  RAJ singh", "meena  patel  "]
cleaned = [clean_name(r) for r in records]
print(cleaned)
```

## Cleaning emails

```python
def clean_email(e):
    return e.strip().lower()

print(clean_email("  User.Name@Domain.COM "))
```

## Cleaning categories - standardized categorical values

```python
def clean_category(c):
    c = c.strip().lower()
    c = c.replace("-", " ").replace("_", " ")
    return c.title()

print(clean_category(" electronics-and_gadgets "))
print(clean_category("HOME_APPLIANCES"))
```

```python
def clean_phone(phone):
    return ''.join(ch for ch in phone if ch.isdigit())

print(clean_phone("+91 98765-43210"))
```

## Interview questions

**Why is string cleaning important in data science?**
Real-world text data is inconsistent — extra spaces, mixed case, different separators for the same category. Uncleaned strings that should be identical (`" Asha "` vs `"asha"`) get treated as different values, which breaks grouping, counting, and joining operations. Cleaning standardizes the data before any analysis runs on it.

**What does strip() do?**
It removes leading and trailing whitespace (spaces, tabs, newlines) from a string, but leaves whitespace in the middle untouched. `"  hi there  ".strip()` gives `"hi there"`.

**How can you convert a string to lowercase?**
With the `.lower()` method: `"HELLO".lower()` returns `"hello"`. It's commonly combined with `.strip()` when standardizing text so comparisons and grouping aren't broken by case differences.
